```python
#2.extract alcohol and citric acid as vectors
alcohol=df["alcohol"].values
citric_acid=df["citric acid"].values
print(alcohol[:10])
print(citric_acid[:10])
```

```python
#3.calculate convariance matrix for and density
X=df[["alcohol","density"]].values
cov_matrix=np.cov(X.T)
print(cov_matrix)
```

```python
#4.eigenvalues and eigenvectors
eigenvalues,eigenvectors=np.linalg.eig(cov_matrix)
print("Eigenvalues:")
print(eigenvalues)
print("Eigenvector:")
print(eigenvectors)
```

```python
#5.most common wine quality and distribution
quality_counts=df["quality"].value_counts().sort_index()
print(quality_counts)
print("Most common quality:")
print(df["quality"].mode()[0])
```
